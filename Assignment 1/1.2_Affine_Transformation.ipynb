```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the input image
img = cv2.imread("/content/image1.jpg")
image = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get image dimensions
height, width = image.shape[:2]

# Translation
M_trans = np.float32([[1, 0, 100], [0, 1, 50]])
translated = cv2.warpAffine(image, M_trans, (width, height))

# Rotation
M_rot = cv2.getRotationMatrix2D((width // 2, height // 2), 45, 1)
rotated = cv2.warpAffine(image, M_rot, (width, height))

# Scaling
scaling_matrix = np.float32([[0.7, 0, 0],[0, 0.7, 0]])
scaled = cv2.warpAffine(image, scaling_matrix, (width, height))

# Display
titles = ["Original", "Translated", "Rotated", "Scaled"]
images = [image, translated, rotated, scaled]

plt.figure(figsize=(12, 8))
for i in range(4):
    plt.subplot(2, 2, i + 1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("off")
plt.tight_layout()
plt.show()
```
