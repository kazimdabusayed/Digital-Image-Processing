```python
import cv2
import matplotlib.pyplot as plt

# Read the input image
image = cv2.imread("image1.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Resize
nearest = cv2.resize(image, None, fx= 3, fy= 3, interpolation=cv2.INTER_NEAREST)
bilinear = cv2.resize(image, None, fx= 3, fy= 3, interpolation=cv2.INTER_LINEAR)
bicubic = cv2.resize(image, None, fx= 3, fy= 3, interpolation=cv2.INTER_CUBIC)

# Display
titles = ["Original", "Nearest", "Bilinear", "Bicubic"]
images = [image, nearest, bilinear, bicubic]

plt.figure(figsize=(12, 8))
for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("off")
plt.tight_layout()
plt.show()
```
